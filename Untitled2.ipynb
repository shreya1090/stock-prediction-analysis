```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
import matplotlib.pyplot as plt
import pickle
import os

# ===== SELECT STOCKS TO TRAIN =====
# Choose which stocks you want to train models for
stocks_to_train = [
    'RELIANCE',
    'TCS',
    'INFY',
    'HDFCBANK',
    'TATAMOTORS'
]

# OR train on ALL stocks in the dataset (will take longer)
# stocks_to_train = [f.replace('.csv', '') for f in os.listdir('/content/dataset/')
#                    if f.endswith('.csv') and f != 'NIFTY50_all.csv' and f != 'stock_metadata.csv']

print(f"Will train models for {len(stocks_to_train)} stocks\n")

# ===== TRAINING FUNCTION =====
def train_stock_model(stock_name):
    print(f"\n{'='*60}")
    print(f"Training model for: {stock_name}")
    print(f"{'='*60}")

    # Load data
    stock_file = f"/content/dataset/{stock_name}.csv"
    if not os.path.exists(stock_file):
        print(f"❌ File not found: {stock_file}")
        return None

    data = pd.read_csv(stock_file)
    data['Date'] = pd.to_datetime(data['Date'])
    data = data.sort_values('Date')
    data.set_index('Date', inplace=True)
    data.ffill(inplace=True)

    print(f"Data points: {len(data)}")
    print(f"Price range: ₹{data['Close'].min():.2f} - ₹{data['Close'].max():.2f}")
    print(f"Latest price: ₹{data['Close'].iloc[-1]:.2f}")

    # Train/Test split
    train_size = int(len(data) * 0.8)
    train_data = data['Close'][:train_size].values.reshape(-1, 1)
    test_data = data['Close'][train_size:].values.reshape(-1, 1)

    # Scale data
    scaler = MinMaxScaler()
    train_scaled = scaler.fit_transform(train_data)
    test_scaled = scaler.transform(test_data)

    # Create sequences
    SEQ_LEN = 60

    def create_sequences(data, seq_len):
        X, y = [], []
        for i in range(seq_len, len(data)):
            X.append(data[i-seq_len:i])
            y.append(data[i])
        return np.array(X), np.array(y)

    X_train, y_train = create_sequences(train_scaled, SEQ_LEN)
    X_test, y_test = create_sequences(test_scaled, SEQ_LEN)

    if len(X_train) < 100:
        print(f"⚠️  Not enough data for {stock_name}. Skipping...")
        return None

    print(f"Training sequences: {X_train.shape[0]}")

    # Build model
    model = Sequential([
        LSTM(50, return_sequences=True, input_shape=(SEQ_LEN, 1)),
        Dropout(0.2),
        LSTM(50, return_sequences=False),
        Dropout(0.2),
        Dense(25),
        Dense(1)
    ])

    model.compile(optimizer='adam', loss='mse')

    # Train
    history = model.fit(
        X_train, y_train,
        validation_data=(X_test, y_test),
        epochs=15,  # Reduced for faster training when doing multiple stocks
        batch_size=32,
        verbose=0
    )

    # Evaluate
    predictions_scaled = model.predict(X_test, verbose=0)
    predictions = scaler.inverse_transform(predictions_scaled)
    y_test_actual = scaler.inverse_transform(y_test)

    rmse = np.sqrt(np.mean((predictions - y_test_actual)**2))
    mae = np.mean(np.abs(predictions - y_test_actual))
    error_pct = (rmse/y_test_actual.mean())*100

    print(f"\n📊 Performance Metrics:")
    print(f"   RMSE: ₹{rmse:.2f}")
    print(f"   MAE: ₹{mae:.2f}")
    print(f"   Error: {error_pct:.2f}%")

    # Predict next price
    last_60_days = scaler.transform(data['Close'].values[-SEQ_LEN:].reshape(-1, 1))
    last_60_days = last_60_days.reshape(1, SEQ_LEN, 1)
    next_price_scaled = model.predict(last_60_days, verbose=0)
    next_price = scaler.inverse_transform(next_price_scaled)

    current_price = data['Close'].iloc[-1]
    predicted_price = next_price[0][0]
    change = predicted_price - current_price

    print(f"\n🎯 Prediction:")
    print(f"   Current: ₹{current_price:.2f}")
    print(f"   Predicted: ₹{predicted_price:.2f}")
    print(f"   Change: ₹{change:.2f} ({(change/current_price)*100:.2f}%)")

    # Save model and scaler
    model_dir = "/content/drive/MyDrive/stock_models"
    os.makedirs(model_dir, exist_ok=True)

    model.save(f"{model_dir}/{stock_name}_model.keras")
    with open(f"{model_dir}/{stock_name}_scaler.pkl", 'wb') as f:
        pickle.dump(scaler, f)

    print(f"✅ Model saved: {stock_name}_model.keras")

    return {
        'stock': stock_name,
        'rmse': rmse,
        'mae': mae,
        'error_pct': error_pct,
        'current_price': current_price,
        'predicted_price': predicted_price,
        'change': change,
        'history': history
    }

# ===== TRAIN ALL SELECTED STOCKS =====
results = []

for stock in stocks_to_train:
    try:
        result = train_stock_model(stock)
        if result:
            results.append(result)
    except Exception as e:
        print(f"❌ Error training {stock}: {str(e)}")

# ===== SUMMARY =====
print("\n" + "="*60)
print("TRAINING SUMMARY")
print("="*60)

if results:
    summary_df = pd.DataFrame(results)
    print(summary_df[['stock', 'current_price', 'predicted_price', 'change', 'error_pct']].to_string(index=False))

    print(f"\n✅ Successfully trained {len(results)} models!")
    print(f"📁 Models saved in: /content/drive/MyDrive/stock_models/")
else:
    print("❌ No models were trained successfully")

# ===== PLOT COMPARISON (for first 3 stocks) =====
if len(results) > 0:
    fig, axes = plt.subplots(min(3, len(results)), 1, figsize=(12, 4*min(3, len(results))))
    if len(results) == 1:
        axes = [axes]

    for idx, result in enumerate(results[:3]):
        ax = axes[idx]
        history = result['history']
        ax.plot(history.history['loss'], label='Train Loss')
        ax.plot(history.history['val_loss'], label='Val Loss')
        ax.set_title(f"{result['stock']} - Training Loss")
        ax.set_xlabel('Epoch')
        ax.set_ylabel('Loss')
        ax.legend()

    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
import matplotlib.pyplot as plt
import pickle
import os

# ===== AUTO-DETECT ALL STOCK CSV FILES =====
# UPDATE THIS PATH TO WHERE YOUR FILES ARE:
dataset_path = '/content/drive/MyDrive/archive (8)/'  # ← Change this if needed

# Check if folder exists
if not os.path.exists(dataset_path):
    print(f"❌ ERROR: {dataset_path} doesn't exist!")
    print("\nAvailable folders in /content/:")
    print(os.listdir('/content/'))
    raise Exception("Dataset folder not found!")

# Get all CSV files
all_files = os.listdir(dataset_path)
csv_files = [f for f in all_files if f.endswith('.csv')]

# Exclude the combined file and metadata
exclude_files = ['NIFTY50_all.csv', 'stock_metadata.csv', 'nifty50_all.csv']
stock_files = [f for f in csv_files if f not in exclude_files]

print(f"📊 Found {len(stock_files)} stock CSV files")
print(f"📁 Dataset location: {dataset_path}\n")
print("Files to train on:")
for i, f in enumerate(stock_files[:10], 1):
    print(f"  {i}. {f}")
if len(stock_files) > 10:
    print(f"  ... and {len(stock_files) - 10} more")

# Ask for confirmation
print(f"\n⚠️  Will train models for ALL {len(stock_files)} stocks")
print("This may take 30-60 minutes depending on your GPU.\n")

# ===== TRAINING FUNCTION =====
def train_stock_model(csv_filename):
    stock_name = csv_filename.replace('.csv', '')
    print(f"\n{'='*60}")
    print(f"Training: {stock_name}")
    print(f"{'='*60}")

    # Load data
    stock_file = os.path.join(dataset_path, csv_filename)

    try:
        data = pd.read_csv(stock_file)
    except Exception as e:
        print(f"❌ Error reading file: {e}")
        return None

    # Check if 'Close' column exists
    if 'Close' not in data.columns:
        print(f"❌ 'Close' column not found. Available columns: {data.columns.tolist()}")
        return None

    data['Date'] = pd.to_datetime(data['Date'])
    data = data.sort_values('Date')
    data.set_index('Date', inplace=True)
    data.ffill(inplace=True)

    print(f"Data points: {len(data)}")
    print(f"Date range: {data.index[0].strftime('%Y-%m-%d')} to {data.index[-1].strftime('%Y-%m-%d')}")
    print(f"Price range: ₹{data['Close'].min():.2f} - ₹{data['Close'].max():.2f}")
    print(f"Latest price: ₹{data['Close'].iloc[-1]:.2f}")

    # Need at least 200 data points for meaningful training
    if len(data) < 200:
        print(f"⚠️  Not enough data ({len(data)} points). Skipping...")
        return None

    # Train/Test split
    train_size = int(len(data) * 0.8)
    train_data = data['Close'][:train_size].values.reshape(-1, 1)
    test_data = data['Close'][train_size:].values.reshape(-1, 1)

    # Scale data
    scaler = MinMaxScaler()
    train_scaled = scaler.fit_transform(train_data)
    test_scaled = scaler.transform(test_data)

    # Create sequences
    SEQ_LEN = 60

    def create_sequences(data, seq_len):
        X, y = [], []
        for i in range(seq_len, len(data)):
            X.append(data[i-seq_len:i])
            y.append(data[i])
        return np.array(X), np.array(y)

    X_train, y_train = create_sequences(train_scaled, SEQ_LEN)
    X_test, y_test = create_sequences(test_scaled, SEQ_LEN)

    if len(X_train) < 50:
        print(f"⚠️  Not enough training sequences. Skipping...")
        return None

    print(f"Training sequences: {X_train.shape[0]}, Test sequences: {X_test.shape[0]}")

    # Build model
    model = Sequential([
        LSTM(50, return_sequences=True, input_shape=(SEQ_LEN, 1)),
        Dropout(0.2),
        LSTM(50, return_sequences=False),
        Dropout(0.2),
        Dense(25),
        Dense(1)
    ])

    model.compile(optimizer='adam', loss='mse')

    # Train (with progress bar)
    print("Training... ", end='')
    history = model.fit(
        X_train, y_train,
        validation_data=(X_test, y_test),
        epochs=15,
        batch_size=32,
        verbose=0
    )
    print("✓")

    # Evaluate
    predictions_scaled = model.predict(X_test, verbose=0)
    predictions = scaler.inverse_transform(predictions_scaled)
    y_test_actual = scaler.inverse_transform(y_test)

    rmse = np.sqrt(np.mean((predictions - y_test_actual)**2))
    mae = np.mean(np.abs(predictions - y_test_actual))
    error_pct = (rmse/y_test_actual.mean())*100

    print(f"📊 RMSE: ₹{rmse:.2f} | MAE: ₹{mae:.2f} | Error: {error_pct:.2f}%")

    # Predict next price
    last_60_days = scaler.transform(data['Close'].values[-SEQ_LEN:].reshape(-1, 1))
    last_60_days = last_60_days.reshape(1, SEQ_LEN, 1)
    next_price_scaled = model.predict(last_60_days, verbose=0)
    next_price = scaler.inverse_transform(next_price_scaled)

    current_price = data['Close'].iloc[-1]
    predicted_price = next_price[0][0]
    change = predicted_price - current_price
    change_pct = (change/current_price)*100

    print(f"🎯 Current: ₹{current_price:.2f} → Predicted: ₹{predicted_price:.2f} ({change_pct:+.2f}%)")

    # Save model and scaler
    model_dir = "/content/drive/MyDrive/stock_models"
    os.makedirs(model_dir, exist_ok=True)

    model.save(f"{model_dir}/{stock_name}_model.keras")
    with open(f"{model_dir}/{stock_name}_scaler.pkl", 'wb') as f:
        pickle.dump(scaler, f)

    print(f"✅ Saved: {stock_name}_model.keras")

    return {
        'stock': stock_name,
        'data_points': len(data),
        'rmse': rmse,
        'mae': mae,
        'error_pct': error_pct,
        'current_price': current_price,
        'predicted_price': predicted_price,
        'change': change,
        'change_pct': change_pct
    }

# ===== TRAIN ALL STOCKS =====
results = []
failed = []

print("\n" + "="*60)
print("STARTING TRAINING")
print("="*60)

for i, csv_file in enumerate(stock_files, 1):
    print(f"\n[{i}/{len(stock_files)}] Processing {csv_file}...")
    try:
        result = train_stock_model(csv_file)
        if result:
            results.append(result)
        else:
            failed.append(csv_file)
    except Exception as e:
        print(f"❌ Error: {str(e)}")
        failed.append(csv_file)

# ===== FINAL SUMMARY =====
print("\n" + "="*60)
print("TRAINING COMPLETE!")
print("="*60)

if results:
    print(f"\n✅ Successfully trained: {len(results)} models")
    print(f"❌ Failed/Skipped: {len(failed)} files")

    # Create summary DataFrame
    summary_df = pd.DataFrame(results)
    summary_df = summary_df.sort_values('change_pct', ascending=False)

    print("\n📊 TOP 10 PREDICTIONS (by % change):")
    print("="*60)
    top_10 = summary_df.head(10)[['stock', 'current_price', 'predicted_price', 'change_pct', 'error_pct']]
    print(top_10.to_string(index=False))

    print("\n📉 BOTTOM 10 PREDICTIONS (by % change):")
    print("="*60)
    bottom_10 = summary_df.tail(10)[['stock', 'current_price', 'predicted_price', 'change_pct', 'error_pct']]
    print(bottom_10.to_string(index=False))

    print("\n📈 STATISTICS:")
    print("="*60)
    print(f"Average Error: {summary_df['error_pct'].mean():.2f}%")
    print(f"Median Error: {summary_df['error_pct'].median():.2f}%")
    print(f"Best Model: {summary_df.loc[summary_df['error_pct'].idxmin(), 'stock']} ({summary_df['error_pct'].min():.2f}% error)")
    print(f"Worst Model: {summary_df.loc[summary_df['error_pct'].idxmax(), 'stock']} ({summary_df['error_pct'].max():.2f}% error)")

    # Save summary to CSV
    summary_csv = "/content/drive/MyDrive/stock_models/training_summary.csv"
    summary_df.to_csv(summary_csv, index=False)
    print(f"\n💾 Summary saved to: {summary_csv}")

    print(f"\n📁 All models saved in: /content/drive/MyDrive/stock_models/")

    # Plot error distribution
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.hist(summary_df['error_pct'], bins=20, edgecolor='black')
    plt.xlabel('Error Percentage (%)')
    plt.ylabel('Number of Stocks')
    plt.title('Distribution of Model Errors')
    plt.axvline(summary_df['error_pct'].mean(), color='r', linestyle='--', label=f'Mean: {summary_df["error_pct"].mean():.2f}%')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.scatter(summary_df['current_price'], summary_df['error_pct'], alpha=0.6)
    plt.xlabel('Current Stock Price (₹)')
    plt.ylabel('Error Percentage (%)')
    plt.title('Error vs Stock Price')

    plt.tight_layout()
    plt.savefig('/content/drive/MyDrive/stock_models/error_analysis.png', dpi=150, bbox_inches='tight')
    plt.show()

    print("📊 Visualization saved: error_analysis.png")

else:
    print("❌ No models were trained successfully")
    if failed:
        print("\nFailed files:")
        for f in failed:
            print(f"  - {f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
import matplotlib.pyplot as plt
import pickle
import os

# ===== AUTO-DETECT ALL STOCK CSV FILES =====
# UPDATE THIS PATH TO WHERE YOUR FILES ARE:
dataset_path = '/content/updated_dataset/'  # ✅ Updated path  # ← Change this if needed

# Check if folder exists
if not os.path.exists(dataset_path):
    print(f"❌ ERROR: {dataset_path} doesn't exist!")
    print("\nAvailable folders in /content/:")
    print(os.listdir('/content/'))
    raise Exception("Dataset folder not found!")

# Get all CSV files
all_files = os.listdir(dataset_path)
csv_files = [f for f in all_files if f.endswith('.csv')]

# Exclude the combined file and metadata
exclude_files = ['NIFTY50_all.csv', 'stock_metadata.csv', 'nifty50_all.csv']
stock_files = [f for f in csv_files if f not in exclude_files]

print(f"📊 Found {len(stock_files)} stock CSV files")
print(f"📁 Dataset location: {dataset_path}\n")
print("Files to train on:")
for i, f in enumerate(stock_files[:10], 1):
    print(f"  {i}. {f}")
if len(stock_files) > 10:
    print(f"  ... and {len(stock_files) - 10} more")

# Ask for confirmation
print(f"\n⚠️  Will train models for ALL {len(stock_files)} stocks")
print("This may take 30-60 minutes depending on your GPU.\n")

# ===== TRAINING FUNCTION =====
def train_stock_model(csv_filename):
    stock_name = csv_filename.replace('.csv', '')
    print(f"\n{'='*60}")
    print(f"Training: {stock_name}")
    print(f"{'='*60}")

    # Load data
    stock_file = os.path.join(dataset_path, csv_filename)

    try:
        data = pd.read_csv(stock_file)
    except Exception as e:
        print(f"❌ Error reading file: {e}")
        return None

    # Check if 'Close' column exists
    if 'Close' not in data.columns:
        print(f"❌ 'Close' column not found. Available columns: {data.columns.tolist()}")
        return None

    data['Date'] = pd.to_datetime(data['Date'])
    data = data.sort_values('Date')
    data.set_index('Date', inplace=True)
    data.ffill(inplace=True)

    print(f"Data points: {len(data)}")
    print(f"Date range: {data.index[0].strftime('%Y-%m-%d')} to {data.index[-1].strftime('%Y-%m-%d')}")
    print(f"Price range: ₹{data['Close'].min():.2f} - ₹{data['Close'].max():.2f}")
    print(f"Latest price: ₹{data['Close'].iloc[-1]:.2f}")

    # Need at least 200 data points for meaningful training
    if len(data) < 200:
        print(f"⚠️  Not enough data ({len(data)} points). Skipping...")
        return None

    # Train/Test split
    train_size = int(len(data) * 0.8)
    train_data = data['Close'][:train_size].values.reshape(-1, 1)
    test_data = data['Close'][train_size:].values.reshape(-1, 1)

    # Scale data
    scaler = MinMaxScaler()
    train_scaled = scaler.fit_transform(train_data)
    test_scaled = scaler.transform(test_data)

    # Create sequences
    SEQ_LEN = 60

    def create_sequences(data, seq_len):
        X, y = [], []
        for i in range(seq_len, len(data)):
            X.append(data[i-seq_len:i])
            y.append(data[i])
        return np.array(X), np.array(y)

    X_train, y_train = create_sequences(train_scaled, SEQ_LEN)
    X_test, y_test = create_sequences(test_scaled, SEQ_LEN)

    if len(X_train) < 50:
        print(f"⚠️  Not enough training sequences. Skipping...")
        return None

    print(f"Training sequences: {X_train.shape[0]}, Test sequences: {X_test.shape[0]}")

    # Build model
    model = Sequential([
        LSTM(50, return_sequences=True, input_shape=(SEQ_LEN, 1)),
        Dropout(0.2),
        LSTM(50, return_sequences=False),
        Dropout(0.2),
        Dense(25),
        Dense(1)
    ])

    model.compile(optimizer='adam', loss='mse')

    # Train (with progress bar)
    print("Training... ", end='')
    history = model.fit(
        X_train, y_train,
        validation_data=(X_test, y_test),
        epochs=15,
        batch_size=32,
        verbose=0
    )
    print("✓")

    # Evaluate
    predictions_scaled = model.predict(X_test, verbose=0)
    predictions = scaler.inverse_transform(predictions_scaled)
    y_test_actual = scaler.inverse_transform(y_test)

    rmse = np.sqrt(np.mean((predictions - y_test_actual)**2))
    mae = np.mean(np.abs(predictions - y_test_actual))
    error_pct = (rmse/y_test_actual.mean())*100

    print(f"📊 RMSE: ₹{rmse:.2f} | MAE: ₹{mae:.2f} | Error: {error_pct:.2f}%")

    # Predict next price
    last_60_days = scaler.transform(data['Close'].values[-SEQ_LEN:].reshape(-1, 1))
    last_60_days = last_60_days.reshape(1, SEQ_LEN, 1)
    next_price_scaled = model.predict(last_60_days, verbose=0)
    next_price = scaler.inverse_transform(next_price_scaled)

    current_price = data['Close'].iloc[-1]
    predicted_price = next_price[0][0]
    change = predicted_price - current_price
    change_pct = (change/current_price)*100

    print(f"🎯 Current: ₹{current_price:.2f} → Predicted: ₹{predicted_price:.2f} ({change_pct:+.2f}%)")

    # Save model and scaler
    model_dir = "/content/drive/MyDrive/stock_models"
    os.makedirs(model_dir, exist_ok=True)

    model.save(f"{model_dir}/{stock_name}_model.keras")
    with open(f"{model_dir}/{stock_name}_scaler.pkl", 'wb') as f:
        pickle.dump(scaler, f)

    print(f"✅ Saved: {stock_name}_model.keras")

    return {
        'stock': stock_name,
        'data_points': len(data),
        'rmse': rmse,
        'mae': mae,
        'error_pct': error_pct,
        'current_price': current_price,
        'predicted_price': predicted_price,
        'change': change,
        'change_pct': change_pct
    }

# ===== TRAIN ALL STOCKS =====
results = []
failed = []

print("\n" + "="*60)
print("STARTING TRAINING")
print("="*60)

for i, csv_file in enumerate(stock_files, 1):
    print(f"\n[{i}/{len(stock_files)}] Processing {csv_file}...")
    try:
        result = train_stock_model(csv_file)
        if result:
            results.append(result)
        else:
            failed.append(csv_file)
    except Exception as e:
        print(f"❌ Error: {str(e)}")
        failed.append(csv_file)

# ===== FINAL SUMMARY =====
print("\n" + "="*60)
print("TRAINING COMPLETE!")
print("="*60)

if results:
    print(f"\n✅ Successfully trained: {len(results)} models")
    print(f"❌ Failed/Skipped: {len(failed)} files")

    # Create summary DataFrame
    summary_df = pd.DataFrame(results)
    summary_df = summary_df.sort_values('change_pct', ascending=False)

    print("\n📊 TOP 10 PREDICTIONS (by % change):")
    print("="*60)
    top_10 = summary_df.head(10)[['stock', 'current_price', 'predicted_price', 'change_pct', 'error_pct']]
    print(top_10.to_string(index=False))

    print("\n📉 BOTTOM 10 PREDICTIONS (by % change):")
    print("="*60)
    bottom_10 = summary_df.tail(10)[['stock', 'current_price', 'predicted_price', 'change_pct', 'error_pct']]
    print(bottom_10.to_string(index=False))

    print("\n📈 STATISTICS:")
    print("="*60)
    print(f"Average Error: {summary_df['error_pct'].mean():.2f}%")
    print(f"Median Error: {summary_df['error_pct'].median():.2f}%")
    print(f"Best Model: {summary_df.loc[summary_df['error_pct'].idxmin(), 'stock']} ({summary_df['error_pct'].min():.2f}% error)")
    print(f"Worst Model: {summary_df.loc[summary_df['error_pct'].idxmax(), 'stock']} ({summary_df['error_pct'].max():.2f}% error)")

    # Save summary to CSV
    summary_csv = "/content/drive/MyDrive/stock_models/training_summary.csv"
    summary_df.to_csv(summary_csv, index=False)
    print(f"\n💾 Summary saved to: {summary_csv}")

    print(f"\n📁 All models saved in: /content/drive/MyDrive/stock_models/")

    # Plot error distribution
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.hist(summary_df['error_pct'], bins=20, edgecolor='black')
    plt.xlabel('Error Percentage (%)')
    plt.ylabel('Number of Stocks')
    plt.title('Distribution of Model Errors')
    plt.axvline(summary_df['error_pct'].mean(), color='r', linestyle='--', label=f'Mean: {summary_df["error_pct"].mean():.2f}%')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.scatter(summary_df['current_price'], summary_df['error_pct'], alpha=0.6)
    plt.xlabel('Current Stock Price (₹)')
    plt.ylabel('Error Percentage (%)')
    plt.title('Error vs Stock Price')

    plt.tight_layout()
    plt.savefig('/content/drive/MyDrive/stock_models/error_analysis.png', dpi=150, bbox_inches='tight')
    plt.show()

    print("📊 Visualization saved: error_analysis.png")

else:
    print("❌ No models were trained successfully")
    if failed:
        print("\nFailed files:")
        for f in failed:
            print(f"  - {f}")
```
